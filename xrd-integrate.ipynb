```python
from pathlib import Path

data_dir = Path(r"G:\My Drive\LPS\20250721_S_MeOMBAI_prestudy_3\XRD")

gfrms = list(data_dir.glob("**/*.gfrm"))

print(gfrms)
```

```python
from collections import defaultdict

# Group files by their parent directory and base name pattern
grouped_files = defaultdict(list)

for file_path in gfrms:
    # Get parent directory name (e.g., "20250721_S_MeO_1")
    parent_dir = file_path.parent.name
    
    # Get filename without extension and remove the last part (xxx)
    filename = file_path.stem
    # Remove the last 3 characters (e.g., "000", "001") to get base pattern
    base_name = filename[:-4]  # Remove "_xxx" part
    
    # Create grouping key
    group_key = f"{base_name}"
    grouped_files[group_key].append(file_path)

# Display the grouped files
for group_key, files in grouped_files.items():
    print(f"{group_key}: {len(files)} files")
    for file in sorted(files):
        print(f"  - {file.name}")
```

```python
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
import numpy as np

# Plot the merged detector
fig, ((ax0, ax1, ax2), axs) = plt.subplots(2, 3, figsize=(16, 10))

for ax, area_detector in zip([ax0, ax1], merger.area_detectors):
    im = ax.imshow(area_detector.image.data, 
                 cmap='viridis',
                 norm=LogNorm(vmin=1, vmax=np.max(area_detector.image.data)),
                 origin='upper')
    ax.set_title('Original Detector Image')
    ax.set_xlabel('Pixel (X)')
    ax.set_ylabel('Pixel (Y)')
    fig.colorbar(im, ax=ax, label='Intensity (counts)')


extent = [
    merger.common_grid['twoth'].min(), 
    merger.common_grid['twoth'].max(),
    merger.common_grid['gamma'].max(), 
    merger.common_grid['gamma'].min()
]

im = ax2.imshow(merger.normalized_data, extent=extent, aspect='auto', norm=LogNorm(vmin=1, vmax=np.nanmax(merger.normalized_data)),
                                  origin='upper')

# 2θ grid lines every 5 degrees
twoth_min, twoth_max = extent[0], extent[1]
for twoth in range(int(twoth_min//5)*5, int(twoth_max//5)*5+5, 5):
    if twoth_min <= twoth <= twoth_max:
        ax2.axvline(x=twoth, color='white', alpha=0.3, linewidth=0.5)

# γ grid lines every 15 degrees
gamma_min, gamma_max = extent[3], extent[2]  # Note: reversed due to imshow
for gamma in range(int(gamma_min//15)*15, int(gamma_max//15)*15+15, 15):
    if gamma_min <= gamma <= gamma_max:
        ax2.axhline(y=gamma, color='white', alpha=0.3, linewidth=0.5)

ax2.set_xlabel('2θ (degrees)')
ax2.set_ylabel('γ (degrees)')
fig.colorbar(im, ax=ax2, label='Intensity', shrink=0.8)

gamma_ranges = [(-120, -60)]  # degrees - ranges to integrate over
twoth_ranges = []
# twoth_ranges = [(5, 7), (11, 13), (17, 19), (24, 26)]  # degrees - ranges to integrate over
colors_list = plt.colormaps.get_cmap("tab10").colors

for i, (gamma_min, gamma_max) in enumerate(gamma_ranges):
    ax2.axhspan(gamma_min, gamma_max, alpha=0.3, color=colors_list[i], 
                      label=f'$\\gamma$ integration: {gamma_min}° to {gamma_max}°')
    
    # Add text labels
    ax2.text(extent[0] + (extent[1] - extent[0]) * 0.02, 
                   (gamma_min + gamma_max) / 2, 
                   f'$\\gamma$: {gamma_min}° to {gamma_max}°',
                   color=colors_list[i], fontweight='bold', fontsize=9,
                   bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))

# Overlay 2θ integration ranges (vertical bands)
for i, (twoth_min, twoth_max) in enumerate(twoth_ranges):
    ax2.axvspan(twoth_min, twoth_max, alpha=0.3, color=colors_list[i], 
                      label=f'2$\\theta$ integration: {twoth_min}° to {twoth_max}°')
    
    # Add text labels
    ax2.text((twoth_min + twoth_max) / 2, 
                   extent[2] + (extent[3] - extent[2]) * 0.02,
                   f'2$\\theta$: {twoth_min}°-{twoth_max}°',
                   color=colors_list[i], fontweight='bold', fontsize=9,
                   rotation=90, ha='center', va='bottom',
                   bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))

import pandas as pd
from scipy import sparse
from scipy.sparse.linalg import spsolve

def baseline_als(y, lam=1e4, p=0.01, niter=10):
    """
    Asymmetric Least Squares baseline correction
    y: input signal
    lam: smoothness parameter (larger = smoother baseline)
    p: asymmetry parameter (0.001 ≤ p ≤ 0.1, smaller = more asymmetric)
    """
    L = len(y)
    D = sparse.diags([1,-2,1],[0,-1,-2], shape=(L,L-2))
    w = np.ones(L)
    for i in range(niter):
        W = sparse.spdiags(w, 0, L, L)
        Z = W + lam * D.dot(D.transpose())
        z = spsolve(Z, w*y)
        w = p * (y > z) + (1-p) * (y < z)
    return z

def simple_baseline_subtraction(x, y, method='polynomial', **kwargs):
    """
    Simple baseline subtraction methods
    """
    if method == 'polynomial':
        degree = kwargs.get('degree', 3)
        # Fit polynomial to the data
        coeffs = np.polyfit(x, y, degree)
        baseline = np.polyval(coeffs, x)
        return y - baseline, baseline
    
    elif method == 'rolling_minimum':
        window = kwargs.get('window', len(y)//20)
        from scipy.ndimage import minimum_filter1d
        baseline = minimum_filter1d(y, size=window, mode='constant')
        return y - baseline, baseline
    
    elif method == 'als':
        baseline = baseline_als(y, **kwargs)
        return y - baseline, baseline
    
    elif method == 'smooth_als':
        # ALS with post-smoothing for extra smooth baseline
        from scipy.signal import savgol_filter
        baseline_raw = baseline_als(y, **kwargs)
        window_length = min(51, len(y)//10)  # Adaptive window size
        if window_length % 2 == 0:  # Must be odd
            window_length += 1
        if window_length < 5:  # Minimum window size
            window_length = 5
        baseline = savgol_filter(baseline_raw, window_length, 3)
        return y - baseline, baseline
    
    else:
        raise ValueError("Method must be 'polynomial', 'rolling_minimum', 'als', or 'smooth_als'")

for i, (gamma_min, gamma_max) in enumerate(gamma_ranges):
    # Find γ indices corresponding to the range
    gamma_mask = (merger.common_grid['gamma'] >= gamma_min) & (merger.common_grid['gamma'] <= gamma_max)
    gamma_indices = np.where(gamma_mask)[0]
    
    if len(gamma_indices) > 0:
        # Integrate (sum) over the γ range
        intensity_slice = np.sum(merger.normalized_data[gamma_indices, :], axis=0)
        
        # Remove baseline - you can choose the method:
        # Method 1: Asymmetric Least Squares with smoothing (recommended for XRD)
        intensity_corrected, baseline = simple_baseline_subtraction(
            merger.common_grid['twoth'], intensity_slice, 
            method='smooth_als', lam=1e9, p=0.00001, niter=20)
        
        # Additional baseline adjustment for log scale plotting
        # Find the minimum value in regions without peaks (background level)
        from scipy.signal import find_peaks
        peaks, _ = find_peaks(intensity_corrected, height=np.max(intensity_corrected)*0.01)
        
        # Create a mask for non-peak regions
        peak_mask = np.zeros_like(intensity_corrected, dtype=bool)
        for peak in peaks:
            # Exclude regions around peaks (±10 points)
            start_idx = max(0, peak - 10)
            end_idx = min(len(intensity_corrected), peak + 11)
            peak_mask[start_idx:end_idx] = True
        
        # Calculate baseline level from non-peak regions
        if np.any(~peak_mask):
            baseline_level = np.percentile(intensity_corrected[~peak_mask], 10)  # 10th percentile
            intensity_corrected = intensity_corrected - baseline_level
        
        # Ensure minimum value is slightly above 0 for log plotting, but keep negative values
        intensity_corrected_log = np.where(intensity_corrected <= 0, 0.1, intensity_corrected)
        
        # Alternative: Simple ALS with higher smoothness
        # intensity_corrected, baseline = simple_baseline_subtraction(
        #     merger.common_grid['twoth'], intensity_slice, 
        #     method='als', lam=1e8, p=0.0001, niter=15)
        
        # Method 2: Polynomial baseline (alternative)
        # intensity_corrected, baseline = simple_baseline_subtraction(
        #     merger.common_grid['twoth'], intensity_slice, 
        #     method='polynomial', degree=3)
        
        # Method 3: Rolling minimum (alternative)
        # intensity_corrected, baseline = simple_baseline_subtraction(
        #     merger.common_grid['twoth'], intensity_slice, 
        #     method='rolling_minimum', window=50)
        
        # Plot both original and baseline-corrected
        axs[0].plot(merger.common_grid['twoth'], intensity_slice, 
                       color=colors_list[i], linewidth=1, alpha=0.5,
                       label=f'$\\gamma$ = {gamma_min}° to {gamma_max}° (raw)')
        axs[0].plot(merger.common_grid['twoth'], intensity_corrected_log, 
                       color=colors_list[i], linewidth=2, 
                       label=f'$\\gamma$ = {gamma_min}° to {gamma_max}° (baseline corrected)')
        axs[0].plot(merger.common_grid['twoth'], baseline, 
                       color=colors_list[i], linewidth=1, linestyle='--', alpha=0.7,
                       label=f'Baseline')
        
        df = pd.DataFrame({
            '2theta': merger.common_grid['twoth'],
            'Intensity_raw': intensity_slice,
            'Intensity_corrected': intensity_corrected,  # Keep original (can be negative)
            'Intensity_corrected_log': intensity_corrected_log,  # For log plotting
            'Baseline': baseline
        })

df.to_csv(Path(data_dir, f'{first_key}.xy'), index=False)

axs[0].set_title('2$\\theta$ Profiles Integrated over $\\gamma$ Ranges')
axs[0].set_xlabel('2$\\theta$ (degrees)')
axs[0].set_ylabel('Integrated Intensity')
axs[0].legend()
axs[0].grid(True, alpha=0.3)
axs[0].set_yscale('log')
# axs[0].set_xlim(5, 30)

# 4. γ Profiles integrated over 2θ ranges
for i, (twoth_min, twoth_max) in enumerate(twoth_ranges):
    # Find 2θ indices corresponding to the range
    twoth_mask = (area_detector.indexes[1] >= twoth_min) & (area_detector.indexes[1] <= twoth_max)
    twoth_indices = np.where(twoth_mask)[0]
    
    if len(twoth_indices) > 0:
        # Integrate (sum) over the 2θ range
        intensity_slice = np.sum(area_detector.data_converted[:, twoth_indices], axis=1)
        
        axs[1].plot(area_detector.indexes[0], intensity_slice, 
                       color=colors_list[i], linewidth=2, 
                       label=f'2$\\theta$ = {twoth_min}° to {twoth_max}° (integrated)')

axs[1].set_title('$\\gamma$ Profiles Integrated over 2$\\theta$ Ranges')
axs[1].set_xlabel('$\\gamma$ (degrees)')
axs[1].set_ylabel('Integrated Intensity')
axs[1].legend()
axs[1].grid(True, alpha=0.3)
axs[1].set_yscale('log')


```
